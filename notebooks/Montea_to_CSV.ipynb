```python
from pathlib import Path

# This notebook is in: thibault-sandbox/notebooks/Testje.ipynb
# So the repo root is one folder up from here
print(Path.cwd().parent)  # run this once to check it prints the right folder
```

```python
from datetime import timezone, timedelta
import pandas as pd

df = pd.read_excel(
    Path.cwd().parent / "data" / "profiles" / "Montea Gent BTM 2.xlsx",
    sheet_name="Sheet1",
    usecols=["dates","con", "gen", "off", "inj"],
    parse_dates=["dates"],
    index_col="dates",
)

# label the index as fixed CET (UTC+1, no DST), then convert to UTC
df.index = (
    df.index
    .tz_localize(timezone(timedelta(hours=1)))
    .tz_convert("UTC")
)

print(df)

```

```python

df.to_csv(Path.cwd().parent / "data" / "csvs" / "Montea Gent BTM 2_UTC.csv") #makes the csv file in the data/profiles folder

```
