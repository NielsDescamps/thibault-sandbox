```python
from pathlib import Path
```

import csv and isolate first five days

```python
import pandas as pd


df = pd.read_csv(Path.cwd().parent / "data" / "csvs" /"merged.csv", parse_dates=["dates"])
df = df.sort_values("dates")

start = df["dates"].min()
end = start + pd.Timedelta(days=5)

first_5_days = df[(df["dates"] >= start) & (df["dates"] < end)]
first_5_days.to_csv(Path.cwd().parent / "data" / "csvs" / "first_5_days_carmeuse_and_da_belgium.csv", index=False)

print(first_5_days.head())
print(first_5_days.tail())
print(f"Rows: {len(first_5_days)}")
```

plot data of first five days

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

fig = make_subplots(
    rows=3, cols=1,
    shared_xaxes=True,
    vertical_spacing=0.06,
    subplot_titles=("con / gen / off / inj [kW]", "price [€/MWh]", "cleared volume [MW]"),
)

for col in ["con", "gen", "off", "inj"]:
    fig.add_trace(go.Scatter(x=first_5_days["dates"], y=first_5_days[col], mode="lines", name=col), row=1, col=1)

price_col = [c for c in first_5_days.columns if "price" in c.lower()][0]
fig.add_trace(go.Scatter(x=first_5_days["dates"], y=first_5_days[price_col], mode="lines", name="price"), row=2, col=1)

volume_col = [c for c in first_5_days.columns if "volume" in c.lower()][0]
fig.add_trace(go.Scatter(x=first_5_days["dates"], y=first_5_days[volume_col], mode="lines", name="cleared volume"), row=3, col=1)

fig.update_layout(
    height=800,
    title="First 5 days",
    hovermode="x unified",
)
fig.update_xaxes(title_text="Date (UTC)", row=3, col=1)

fig.show(renderer="browser")  # swap to fig.show() if inline rendering is working now
```
